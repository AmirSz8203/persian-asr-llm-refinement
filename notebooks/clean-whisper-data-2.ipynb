```python
import pandas as pd

raw_pairs_df = pd.read_csv("/kaggle/input/datasets/amirsafarzadeh8203/raw-pairs-v10k/raw_pairs_v10k.csv")
print("تعداد کل ردیف‌ها:", len(raw_pairs_df))
print(raw_pairs_df.nunique())
```

```python
pair_dupes = raw_pairs_df.duplicated(subset=["sentence", "raw_whisper"], keep=False)
print("تعداد ردیف‌هایی که جفت (sentence, raw_whisper) تکراریه:", pair_dupes.sum())

raw_pairs_df[pair_dupes].sort_values(["sentence", "raw_whisper"]).head(20)
```

```python
before = len(raw_pairs_df)
raw_pairs_df = raw_pairs_df.drop_duplicates(subset=["sentence", "raw_whisper"], keep="first")
after = len(raw_pairs_df)
print(f"{before} -> {after} ({before - after} ردیف duplicate واقعی حذف شد)")
```

```python
print("تعداد نهایی ردیف‌ها:", len(raw_pairs_df))
print(raw_pairs_df.nunique())
# اینجا طبیعیه که هنوز unique sentence از unique raw_whisper کمتر باشه
# چون یک جمله می‌تونه چند raw_whisper متفاوت (از گوینده‌های مختلف) داشته باشه — این یه باگ نیست
```

```python
dup_whisper = raw_pairs_df[raw_pairs_df["raw_whisper"].duplicated(keep=False)]
dup_whisper.sort_values("raw_whisper")
```

```python
whisper_dupes = raw_pairs_df["raw_whisper"].duplicated(keep=False)
print("تعداد ردیف‌هایی که raw_whisper مشترک با ردیف دیگه‌ای داره:", whisper_dupes.sum())

raw_pairs_df[whisper_dupes].sort_values("raw_whisper")
```

```python
before = len(raw_pairs_df)
raw_pairs_df = raw_pairs_df[~raw_pairs_df["raw_whisper"].duplicated(keep=False)]
after = len(raw_pairs_df)
print(f"{before} -> {after} ({before - after} ردیف حذف شد)")
```

```python
raw_pairs_df = raw_pairs_df.reset_index(drop=True)
print("تعداد نهایی ردیف‌ها:", len(raw_pairs_df))
print(raw_pairs_df.nunique())
```

```python
raw_pairs_df.to_csv("/kaggle/working/raw_pairs_v10k_cleaned.csv", index=False)
print("ذخیره شد:", raw_pairs_df.shape)
```
